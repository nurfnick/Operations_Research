```python
def timeValueOfMoney(years):
  if years == 0:
    return 0.005
  else:
    return timeValueOfMoney(years-1)*(1+0.0327)
```

```python
timeValueOfMoney(years)
```

```python
def fib(n):
  if n == 0:
    return 1
  elif n==1:
    return 1
  else:
    return fib(n-1)+fib(n-2)

fib(5)
```

```python
fib(100)
```

```python
[fib(n) for n in range(20)]
```

```python
def stupid(n):
  if n==0:
    return 1
  elif n== 1:
    return 2
  else:
    return 3*stupid(n-1)+4*stupid(n-2)

[stupid(n) for n in range(20)]
```

```python
def tradstupid(n):
  return 3/5*4**n+2/5*(-1)**n
```

```python
[tradstupid(n) for n in range(20)]
```
