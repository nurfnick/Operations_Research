## Memoization

```python
def amemo(n):
  if n==0:
    return 1
  elif n==1:
    return 1
  else:
    l= [0]*(n+1)
    l[0] = 1
    l[1] = 1
    for i in range(2,n+1):
      l[i] = l[i-1]+6*l[i-2]+3*i
    return l[-1]

amemo(5)
```
