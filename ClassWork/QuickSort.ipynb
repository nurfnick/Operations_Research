```python
import random

def quicksort_random_pivot(arr):
    """
    Sorts a list using the Quicksort algorithm with a randomly chosen pivot.
    """
    if len(arr) <= 1:
        return arr

    # Choose a random pivot from the array
    pivot = random.choice(arr)

    less = [x for x in arr if x < pivot]
    equal = [x for x in arr if x == pivot]
    greater = [x for x in arr if x > pivot]

    return quicksort_random_pivot(less) + equal + quicksort_random_pivot(greater)

# Example usage:
my_list = [3, 6, 8, 10, 1, 2, 1]
sorted_list = quicksort_random_pivot(my_list)
print(f"Original list: {my_list}")
print(f"Sorted list: {sorted_list}")
```

```python
def quicksort_middle_pivot(arr):
    """
    Sorts a list using the Quicksort algorithm with a randomly chosen pivot.
    """
    if len(arr) <= 1:
        return arr

    # Choose a random pivot from the array
    pivot = arr[len(arr)//2]

    less = [x for x in arr if x < pivot]
    equal = [x for x in arr if x == pivot]
    greater = [x for x in arr if x > pivot]

    return quicksort_middle_pivot(less) + equal + quicksort_middle_pivot(greater)

# Example usage:
my_list = [3, 6, 8, 10, 1, 2, 1,3,7,100,42]
sorted_list = quicksort_random_pivot(my_list)
print(f"Original list: {my_list}")
print(f"Sorted list: {sorted_list}")
```
