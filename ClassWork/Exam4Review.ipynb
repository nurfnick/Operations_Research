## 1

C(150,10) = \binom{150}{10}

## 2

$$
C(20,3)\times C(130,7) = \binom{20}3\binom{130}7

## 3

At least 1 defective? Or compute the probability of no defectives and take from total

$$
1-\frac{C(130,10)}{C(150,10)} = 1-\frac{\binom{130}{10}}{\binom{150}{10}}
$$

```python
1-math.comb(130,10)/math.comb(150,10)
```

## 5

We discussed...binomial distribution!

$$
P(x=k) = \binom nk p^k (1-p)^{n-k}
$$

Here $p = 0.10$, $n = 150$, and $k=20$

$$P = \binom{150}{20}0.1^{20}0.9^{130}$$

```python
math.comb(150,20)*0.1**20*0.9**130
```

## 7 Repeat 6 with code!

```python
import random


def_count = 0
for i in range(150):
  if random.random() < 0.1:
    def_count += 1
```

```python
print(def_count)
```

```python
l = []

for j in range(1000):
  def_count = 0
  for i in range(150):
    if random.random() < 0.1:
      def_count += 1
  l.append(def_count)
```

```python
sum(l)/1000
```

```python
import matplotlib.pyplot as plt

plt.hist(l)
```
