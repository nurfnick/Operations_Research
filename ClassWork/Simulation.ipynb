```python
def fib(n):
  if n <= 0:
    return 1
  elif n == 1:
    return 1
  else:
    return fib(n-1) + fib(n-2)
```

```python
l = []
currentpot = []
pot = 500
maxbet = 1
loss = 2
for i in range(100):
  bet = fib(loss-2)
  loss = 0
  while (random.random()<0.54) & (pot > bet):
    loss += 1
    pot -= bet
    bet = fib(loss)
    maxbet = max(bet,maxbet)
    currentpot.append(pot)
  pot += bet
  l.append(loss)
  currentpot.append(pot)


l
pot
```

```python
l = []
currentpot = []
pot = 500
maxbet = 1
for i in range(100):
  loss = 0
  bet = 1
  while (random.random()<0.6) & (pot > bet):
    loss += 1
    pot -= bet
    bet *= 2
    maxbet = max(bet,maxbet)
  pot += bet
  l.append(loss)
  currentpot.append(pot)
```
